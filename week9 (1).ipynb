## **Activity 1** Scenario: Perceptron for Basic Logic (AND Gate) You’ll implement a Perceptron to mimic an AND logic gate.


```python
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
#AND gate inputs and outputs

X_and=np.array([[0,0],[0,1],[1,0],[1,1]])
y_and=np.array([0,0,0,1])
```

```python
#initializing weights and biases randomly
np.random.seed(42)
weights=np.random.rand(2)*0.1
bias=np.random.rand(1)*0.1

weights,bias
```

```python
learning_rate=0.1
epochs=20

def step_function(z):
  return 1 if z>=0 else 0

for epoch in range(epochs):
  for x,y_true in zip(X_and,y_and):
    z=np.dot(x,weights)+bias
    y_pred=step_function(z)

    error=y_true-y_pred

    weights+=error*learning_rate*x
    bias+=error*learning_rate
```

```python
weights,bias #upated weights and bias
```

```python
#prediction

for x in X_and:
  z=np.dot(x,weights)+bias
  prediction=step_function(z)
  print(f'Input:{x}  Prediction:{prediction}')
```

## **Activity 2** Scenario: Multilayer Perceptron for Classification You’ll use an MLP to classify basic shapes (e.g., squares vs. circles) represented by simplified features

```python
import pandas as pd
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import numpy as np
```

```python
# Target : 0= Square 1=Circle
X_shapes = np.array([
    [4,1.0],[4,.05],[4,0.95], #squares
    [0,1.0],[0,1.1],[0,0.9],  #circles
    [4,0.5],[0,2.0]           #some noisy points
])

y_shapes=np.array([0,0,0,1,1,1,0,1])
```

```python
X_train_s,X_test_s,y_train_s,y_test_s=train_test_split(
    X_shapes,y_shapes,test_size=0.3,random_state=42
    )
```

```python
X_train_s
```

```python
y_train_s
```

```python
model=MLPClassifier(hidden_layer_sizes=(5,3),max_iter=500,random_state=42,solver='adam')
model.fit(X_train_s,y_train_s)
```

```python
y_pred_s=model.predict(X_test_s)
y_pred_s
```

```python
accuracy_score(y_pred_s,y_test_s)
```

In Backpropagation:



1.   After Forward propagation, the loss is calculated using a loss fungtion e.g: OLS
2.   the los function L is differentiated w.r.t weight and bias and gradient is obtained which is used to adjust them .
3. the derivative of loss function is propagated backwards  using chain rule and is used to calculate the gradient in the previous layer for each weights and biases present.
4. Then the gradients and of subsequent layers are calculated in the same way where they use the chain rule to obtain the gradients of their weights and biases. This happens until we reach the input layer (where we don't have to make any adjustment).

## **Activity 4** Problem 4: Activation Functions Comparison Compare different activation functions(sigmoid, tanh, ReLU).


sigmoid:
1. gives value bw 0(inclusive) and 1(exclusive).
2. can be used in the output layer where probability is to be calculated
3. vanising gradient problem. as the value of z become larger, the gradient decreases
4. require more computation as compared to ReLU

tanh

1. gives values bw -1 and 1 (bot inclusive)
2. 0 centered
3. also has vanising gradient poblem. in large values even more than sigmoid
4. Requires more computation as compare to ReLu

ReLU
1. gives values in the range [0,infinity)
2. no vanising gradient for +ve value of z. no gradient value for -ve or 0 z.
3. computationally cheap
4. Can cause neurons to become irrelevant (Dying ReLU Problem).
