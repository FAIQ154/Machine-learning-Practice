```python
reading=[22.5,23.1,20.9,24,21.5,19.,26.2,23.5]
farenheit_readings=[round(9/5*c,2) for c in reading]
farenheit_readings
```

```python
hot_days=[temp for temp in reading if temp>23]
hot_days
```

```python
print(farenheit_readings,hot_days)
```

```python
log1_events = {'A101', 'B202', 'C303', 'A101', 'D404'}
log2_events = {'C303', 'E505', 'F606', 'B202'}
log1_events=set(list(log1_events))
log1_events
```

```python
common_events=set(log1_events)&set(log2_events)
common_events
```

```python
union_events=set(log1_events)|set(log2_events)
union_events
```
