**Activity 2** Scenario: K-Fold Cross-Validation for Robust Evaluation You want a more reliable
estimate of how your model will perform on new, unseen data, rather than relying on a single
train-test split.


```python
from sklearn.model_selection import KFold,cross_val_score
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression(max_iter=1000)
kf=KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scores=cross_val_score(
    model,
    X,
    y,
    cv=kf,
    scoring='accuracy'
)

for (i , score) in enumerate(scores):
  print(f'Fold{i}:  Score{score:.4f}')
else:
  print('Average Score:',scores.mean())
```

**BENEFITS OF K-FOLD CROSS VALIDATION**


1. model sees all of the data. This can help give better idea of accuracy.

2. Since all the data is used to train the model it can be useful to check the accuracy of a model if dataset is small and splitting into training - testing data is not possible.

3. A model can have a poor result on the testing data of some split giving misleading evaluation. KFold Cross validation gives idea of average split.
