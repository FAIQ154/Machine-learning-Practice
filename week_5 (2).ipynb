# **Activity 2** Scenario: Employee Performance Data - Outlier Detection You are evaluating employee performance data, but some entries might be erroneous or extreme outliers that could skew analysis.

**Activity:**

▶ Identify Outliers (IQR Method): Calculate the first quartile (Q1) and third quartile (Q3) for ’PerformanceScore’. Then calculate the Interquartile Range (IQR = Q3 - Q1).

▶ Define upper bound (Q3 + 1.5 * IQR) and lower bound (Q1 - 1.5 * IQR) for outlier detection.

▶ Identify and print the EmployeeID and PerformanceScore of all entries that are considered outliers based on these bounds.

▶ Strategy (Capping/Winsorization): Replace any PerformanceScore above the upper bound with the upper bound, and any below the lower bound with the lower bound. Print the PerformanceScore column after capping, observing the change in outlier values.

Hint: Use df.quantile(0.25) and df.quantile(0.75) for Q1 and Q3. Use boolean indexing for capping (e.g., df.loc[df[’col’] > upper_bound, ’col’] = upper_bound).

```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
performance_scores=np.random.normal(loc=75,scale=10,size=100) #mean,standerd deviation and no. of values
performance_scores[10]=100
performance_scores[25]=-20
performance_scores[50]=5
performance_scores
```

```python
employee_df=pd.DataFrame ({'EmployeeID': range(1,101),'PerformanceScore':performance_scores.round(0)})
employee_df.head()
```

```python
print('Original Employee Performance Dataset (first 10 rows)')
employee_df.head(10)
```

```python
print('Original Employee Performance Dataset (last 10 rows)')
employee_df.tail(10)
```

### **ACTIVITY:**

```python
q1=employee_df["PerformanceScore"].quantile(0.25)
q3=employee_df["PerformanceScore"].quantile(0.75)
IQR=q3-q1

print(q1)
print(q3)
print(IQR)
```

```python
upper_bound=q3+1.5*IQR
lower_bound=q1-1.5*IQR

print('Upper Bound:',upper_bound)
print('Lower Bound:',lower_bound)
```

```python
outliers=employee_df[(employee_df["PerformanceScore"] > upper_bound) | (employee_df["PerformanceScore"] < lower_bound) ]
outliers
```

```python
employee_df.loc[employee_df["PerformanceScore"]>upper_bound,"PerformanceScore"]=upper_bound
employee_df.loc[employee_df["PerformanceScore"]<lower_bound,"PerformanceScore"]=lower_bound
```

```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(43)
```

```python
temp_data=pd.DataFrame({
    'CityName': ['London', 'Paris', 'New York', 'Tokyo',
'Berlin', 'Rome', 'Sydney', 'Cairo'],
    'AvgTempC': np.random.uniform(10,30,8).round(1),
    'Humidity': np.random.uniform(50,90,8).round(1)
})
print('Temperature Data')
temp_data
```

```python
city_pop=pd.DataFrame({
    'City':['London', 'Paris', 'New York', 'Tokyo',
'Sydney',  'Dubai', 'Beijing'],
    'Country': ['UK','France','USA','Japan','Australia','UAE','Chine'],
    'PopulationMillions': [9, 2.1, 8.4, 14, 5.3, 3.1, 21.5]

})
city_pop
```

```python
merged_df=pd.merge(temp_data,city_pop,left_on='CityName',right_on='City',how='inner')
merged_df
```

## To keep all the temperatures in the dataframe we will do left join as follows. this will copy all the data of the left dataframe irrespective of the corresponding matching key value being available in the right dataframe

```python
merged_df2=pd.merge(temp_data,city_pop,left_on='CityName',right_on='City',how='left')
merged_df2
```
