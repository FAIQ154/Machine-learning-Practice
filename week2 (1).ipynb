```python
class Car:
  def __init__(self,make,model,year):
    self.make=make
    self.model=model
    self.year=year
  def display(self):
    print('make:',self.make)
    print('model:',self.model)
    print('year:',self.year)
```

```python
Toyota_Car=Car('Toyota Camry 2020','Camry New',2000)
Toyota_Car.display()
```

```python
Honda_Car=Car('Honda Civic 2022','Civic New',2022)
Honda_Car.display()
```

```python
class BankAccount:
  def __init__(self,account_number,balance):
    self.account_number=account_number
    self._balance=balance

  def deposit(self,new_amount):
    self._balance_+=new_amount
    print('New_Balance:',self._balance)

  def withdraw(self,amount):
     if amount>self._balance:
       print('Insufficient Amount')
     else:
      self._balance-=amount
      print('Amount Withdrawn')
      print('New Balance:',self._balance)

```

```python
account1 = BankAccount(1,500)
```

```python
account1.withdraw(200 )
```

```python
account1.withdraw(400)
```

```python
class Student:
  def __init__(self,name,student_id):
    self.name=name
    self.student_id=student_id

  def get_details(self):
    print('Name:',self.name)
    print('Student ID',self.student_id)
```

```python
class GraduateStudent(Student):
  def __init__(self,name,student_id,thesis_topic):
    super().__init__(name,student_id)
    self.thesis_topic=thesis_topic

  def get_details(self):
    print('Name:',self.name)
    print('student_id:',self.student_id)
    print('thesis_topic:',self.thesis_topic)
```

```python
student_1=Student('A',1)
student_1.get_details()
```

```python
grad_student1=GraduateStudent('A',1,'Machine Learning')
grad_student1.get_details()
```

```python
import math
class Shape:
  def area(self):
    raise NotImplementedError('Overright the function in other classes')

class Rectangle(Shape):
    def __init__(self,width,height):
      self.width=width
      self.height=height

    def area(self):
      return self.width*self.height

class Circle(Shape):
    def __init__(self,radius):
      self.radius=radius

    def area(self):
      return 2*math.pi*self.radius

```

```python
shapes=[Rectangle(7,8),Circle(42)]
for shape in shapes:
  print(shape.area())
```
