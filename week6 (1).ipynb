# **Activity 2** Scenario: Iris Flower Data - Normalization & Standardization The famous Iris dataset is commonly used to demonstrate classification. Scaling its features is often a good practice








```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
iris=load_iris(as_frame=True)
iris_df=iris.frame
print('first five rows:')
iris_df.head()
```

```python
numerical_df=iris_df.select_dtypes(include='number')
features_df=numerical_df.drop(columns='target')
```

```python
minmaxscalar=MinMaxScaler()
normalized_data=minmaxscalar.fit_transform(features_df)
normalized_df=features_df.copy()
normalized_df[:]=normalized_data
normalized_df.head()
```

```python
standardscalar=StandardScaler()
standardized_data=standardscalar.fit_transform(features_df)
standardized_df=features_df.copy()
standardized_df[:]=standardized_data
standardized_df.head()
```

```python
standardized_df.mean()  #mean is approximately 0
```

```python
standardized_df.std()  #standard deviation is aproximately 1
```

Standardization centeres the data around mean 0 and standard devietion 1 which means while most of the data would lie in this range, it can go beyond it.
While in scaling the data is restricted between o and 1. scaling is highly sensitive to outliers.

a uniformly distributed data can have scaling applied to it while an for an uneven data standardization is prefered

# **Activity 3** Scenario: Adult Income Data - Discretization for Categorization The ”Adult” dataset contains demographic information. The ’Age’ feature is continuous, but for some analyses, it might be more useful to group ages into categories (bins).

```python
import pandas as pd
from google.colab import files
uploaded=files.upload()
```

```python
column_names=['age','workclass','fnlwgt','education','educationnum','marital-status','occupation','relationship','race',
              'sex','capital-gain','capital-loss','hours-per-week','native-country','income']

adult_df=pd.read_csv('adult.data',header=None,names=column_names,skipinitialspace=True)
#by default , the fn will take the first row as header . not suitable for this scenario
adult_df.head()
```

```python
adult_df['age'].min()
```

```python
adult_df['age'].max()
```

```python
bins=[17,30,45,60,adult_df['age'].max()]
labels=['Young Adult','Middle Aged','Senior Adult','Elderly']
adult_df['age group']=pd.cut(adult_df['age'],bins=bins,labels=labels,include_lowest=True)
adult_df['age group'].value_counts()
```

**Discretization can be important when:**

1.   **Interpretabiity is Important**
2.   **Large no. of values may lead to the model catch noise. discretization can help in better generalization**
3. **Exact Numerical Value isn't very meaningful**







# **Activity 4** Challenging Activity: MovieLens Dataset - Feature Engineering & Aggregation (Data Reduction/Transformation) You’re working with movie rating data and want to create a reduced,more insightful dataset of movie statistics.


```python
import pandas as pd
from google.colab import files
uploaded=files.upload()
```

```python
ratings_df=pd.read_csv('u.data',sep='\t',
                       header=None,
                       names=[
                           'user_id',
                           'item_id',
                           'rating',
                           'timestamp'
                       ]
                       )
ratings_df.head()
```

```python
movies_df=pd.read_csv('u.item',sep='|',
                      encoding='latin-1',
                      header=None,
                      usecols=[0,1],
                      names=['movie id','movie title'])
movies_df.head()
```

```python
movie_stats=ratings_df.groupby('item_id').agg(average_ratings=('rating','mean'),
                           total_ratings=('rating','count')).reset_index()
movie_stats.head()
```

```python
merge_df=pd.merge(movies_df,movie_stats,left_on='movie id',right_on='item_id',how='inner')
merge_df.head()
```

```python
movie_stats_df=merge_df[['movie title','average_ratings','total_ratings']]
movie_stats_df.head()
```

```python
movie_stats_df=movie_stats_df.sort_values(by='total_ratings',ascending=False)
movie_stats_df.head()
```
